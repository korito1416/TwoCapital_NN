```python


def load_model(params, v_nn_checkpoint_path, i_g_nn_checkpoint_path, i_d_nn_checkpoint_path):
    test_model = model.model(params)
    
    n_inputs = 7 # if "post_tech_post_damage" in params["model_type"] else 8
    test_model.v_nn.build((params["batch_size"], n_inputs))
    test_model.i_g_nn.build((params["batch_size"], n_inputs))
    test_model.i_d_nn.build((params["batch_size"], n_inputs)) 
    
    test_model.v_nn.load_weights(v_nn_checkpoint_path)
    test_model.i_g_nn.load_weights(i_g_nn_checkpoint_path)
    test_model.i_d_nn.load_weights(i_d_nn_checkpoint_path) 

    return test_model


reloaded_model = load_model(params, v_nn_checkpoint_path, i_g_nn_checkpoint_path, i_d_nn_checkpoint_path )
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

def plot_V_and_dVdY(reloaded_model, gamma_3_values=[0.1, 0.2], A_g_prime=0.12, log_xi=-1.0):
    # Define ranges for Y and logK
    Y_values = np.linspace(0, 4, 201)
    logK_values = np.linspace(4.0, 7.0, 201)  # Adjust as needed
    R_values = np.linspace(0.01, 0.99, 20)     # 20 evenly spaced values for R

    # Create meshgrid for Y and logK
    Y_grid, logK_grid = np.meshgrid(Y_values, logK_values)

    # Flatten the grids for vectorized computations
    Y_flat = Y_grid.flatten()
    logK_flat = logK_grid.flatten()

    # Prepare constant arrays
    A_g_prime_array = np.full_like(Y_flat, A_g_prime)
    log_xi_array = np.full_like(Y_flat, log_xi)

    # Colormaps for different gamma_3 values
    colormaps = ['viridis', 'plasma']

    # Loop over each R value
    for idx, R in enumerate(R_values):
        print(f"Processing R = {R:.2f} ({idx+1}/{len(R_values)})")
        R_array = np.full_like(Y_flat, R)

        V_grids = {}
        dV_dY_grids = {}

        for i, gamma_3 in enumerate(gamma_3_values):
            gamma_3_array = np.full_like(Y_flat, gamma_3)

            # Convert arrays to tensors
            Y_var = tf.Variable(Y_flat, dtype=tf.float32)
            logK_tensor = tf.constant(logK_flat, dtype=tf.float32)
            R_tensor = tf.constant(R_array, dtype=tf.float32)
            gamma_3_tensor = tf.constant(gamma_3_array, dtype=tf.float32)
            A_g_prime_tensor = tf.constant(A_g_prime_array, dtype=tf.float32)
            log_xi_tensor = tf.constant(log_xi_array, dtype=tf.float32)

            # Use GradientTape to compute V and dV/dY
            with tf.GradientTape() as tape:
                tape.watch(Y_var)
                # Stack inputs for the model
                inputs = tf.stack([
                    logK_tensor, 
                    R_tensor, 
                    Y_var, 
                    gamma_3_tensor, 
                    A_g_prime_tensor, 
                    log_xi_tensor, 
                    log_xi_tensor
                ], axis=1)

                # Predict V using the model
                V = reloaded_model.v_nn(inputs)
                V = tf.squeeze(V)

            # Compute gradient of V with respect to Y
            dV_dY = tape.gradient(V, Y_var)

            # Reshape outputs to match the grid shape
            V_grid = V.numpy().reshape(Y_grid.shape)
            dV_dY_grid = dV_dY.numpy().reshape(Y_grid.shape)

            V_grids[gamma_3] = V_grid
            dV_dY_grids[gamma_3] = dV_dY_grid

            # Print min and max of dV/dY to check if it's less than zero
            print(f"gamma_3={gamma_3}: dV/dY min: {dV_dY_grid.min():.4f}, max: {dV_dY_grid.max():.4f}")

        # Now plot V as a 3D surface plot, with both gamma_3 values
        fig = plt.figure(figsize=(14, 10))
        ax = fig.add_subplot(111, projection='3d')

        for i, gamma_3 in enumerate(gamma_3_values):
            surf = ax.plot_surface(Y_grid, logK_grid, V_grids[gamma_3],
                                   cmap=colormaps[i], alpha=0.6, edgecolor='none',
                                   label=f'gamma_3={gamma_3}')
        ax.set_xlabel('Y', fontsize=12)
        ax.set_ylabel('logK', fontsize=12)
        ax.set_zlabel('V', fontsize=12)
        ax.set_title(f'V as a function of Y and logK (R={R:.2f})', fontsize=14)
        ax.view_init(elev=30, azim=135)  # Adjust the viewing angle
        # Create custom legend
        from matplotlib.lines import Line2D
        custom_lines = [Line2D([0], [0], color='white', lw=4, markerfacecolor='white',
                               markeredgecolor='black', markersize=0,
                               label=f'gamma_3={gamma_3}', alpha=0) for gamma_3 in gamma_3_values]
        ax.legend(custom_lines, [f'gamma_3={gamma_3}' for gamma_3 in gamma_3_values])
        plt.tight_layout()
        plt.show()

        # Plot V as contour plots
        plt.figure(figsize=(12, 8))
        for i, gamma_3 in enumerate(gamma_3_values):
            contour = plt.contourf(Y_grid, logK_grid, V_grids[gamma_3],
                                   levels=20, cmap=colormaps[i], alpha=0.5)
        cbar = plt.colorbar(contour)
        cbar.ax.set_ylabel('V', fontsize=12)
        plt.xlabel('Y', fontsize=12)
        plt.ylabel('logK', fontsize=12)
        plt.title(f'Heatmap of V (R={R:.2f})', fontsize=14)
        plt.legend([f'gamma_3={gamma_3}' for gamma_3 in gamma_3_values])
        plt.tight_layout()
        plt.show()

        # Plot dV/dY as a 3D surface plot
        fig = plt.figure(figsize=(14, 10))
        ax = fig.add_subplot(111, projection='3d')

        for i, gamma_3 in enumerate(gamma_3_values):
            surf = ax.plot_surface(Y_grid, logK_grid, dV_dY_grids[gamma_3],
                                   cmap=colormaps[i], alpha=0.6, edgecolor='none',
                                   label=f'gamma_3={gamma_3}')
        ax.set_xlabel('Y', fontsize=12)
        ax.set_ylabel('logK', fontsize=12)
        ax.set_zlabel('dV/dY', fontsize=12)
        ax.set_title(f'dV/dY as a function of Y and logK (R={R:.2f})', fontsize=14)
        ax.view_init(elev=30, azim=135)
        # Custom legend
        ax.legend(custom_lines, [f'gamma_3={gamma_3}' for gamma_3 in gamma_3_values])
        plt.tight_layout()
        plt.show()

        # Plot dV/dY as contour plots
        plt.figure(figsize=(12, 8))
        for i, gamma_3 in enumerate(gamma_3_values):
            contour = plt.contourf(Y_grid, logK_grid, dV_dY_grids[gamma_3],
                                   levels=20, cmap=colormaps[i], alpha=0.5)
        cbar = plt.colorbar(contour)
        cbar.ax.set_ylabel('dV/dY', fontsize=12)
        plt.xlabel('Y', fontsize=12)
        plt.ylabel('logK', fontsize=12)
        plt.title(f'Heatmap of dV/dY (R={R:.2f})', fontsize=14)
        plt.legend([f'gamma_3={gamma_3}' for gamma_3 in gamma_3_values])
        plt.tight_layout()
        plt.show()

        print("\n")
```
