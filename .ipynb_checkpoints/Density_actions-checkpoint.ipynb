# $\xi=0.05$

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# ---------- CONFIG ----------
base_folder = "/project/lhansen/Cap_damage/TwoStageTechJump_SITE_Pretrain/SimulationResults/paths_ξ_0.05"
seed_min, seed_max = 1, 100
save_fig = True
STRICT_SAME_STAGE = True
# ----------------------------

Ir_t, Ir_t1 = [], []
Id_t, Id_t1 = [], []
Ig_t, Ig_t1 = [], []

stages = {0: "Pre-Tech", 1: "Interm-Tech", 2: "Post-Tech"}
Ir_stage = {s: {"t": [], "t1": []} for s in stages}
Id_stage = {s: {"t": [], "t1": []} for s in stages}
Ig_stage = {s: {"t": [], "t1": []} for s in stages}

def load_arrays(folder, seed):
    D   = np.load(os.path.join(folder, f"damage_state_array_{seed}.npy"))
    T   = np.load(os.path.join(folder, f"tech_state_array_{seed}.npy"))
    Z   = np.load(os.path.join(folder, f"Z_array_{seed}.npy"))
    Ir  = np.load(os.path.join(folder, f"I_r_array_{seed}.npy"))
    Id  = np.load(os.path.join(folder, f"I_d_array_{seed}.npy"))
    Ig  = np.load(os.path.join(folder, f"I_g_array_{seed}.npy"))
    return D, T, Z, Ir, Id, Ig

for seed in range(seed_min, seed_max + 1):
    try:
        D, T, Z, Ir, Id, Ig = load_arrays(base_folder, seed)
    except FileNotFoundError:
        print(f"[Seed {seed:03d}] missing files, skipping.")
        continue

    n_paths, T1 = D.shape
    for i in range(n_paths):
        d = D[i]
        trans = np.where((d[:-1] == 0) & (d[1:] == 1))[0]
        if trans.size == 0:
            continue
        t = trans[0] + 1
        if t < 0 or t + 1 >= T1:
            continue

        # Adjust I_d and I_g by Z
        z_t, z_t1 = Z[i, t], Z[i, t + 1]
        ir_t, ir_t1 = Ir[i, t], Ir[i, t + 1]
        id_t, id_t1 = Id[i, t] * (1 - z_t), Id[i, t + 1] * (1 - z_t1)
        ig_t, ig_t1 = Ig[i, t] * z_t, Ig[i, t + 1] * z_t1

        if np.isfinite(ir_t): Ir_t.append(ir_t)
        if np.isfinite(ir_t1): Ir_t1.append(ir_t1)
        if np.isfinite(id_t): Id_t.append(id_t)
        if np.isfinite(id_t1): Id_t1.append(id_t1)
        if np.isfinite(ig_t): Ig_t.append(ig_t)
        if np.isfinite(ig_t1): Ig_t1.append(ig_t1)

        tb, ta = int(T[i, t]), int(T[i, t + 1])
        if tb in stages and ta in stages and ((not STRICT_SAME_STAGE) or (tb == ta)):
            s = tb
            if np.isfinite(ir_t): Ir_stage[s]["t"].append(ir_t)
            if np.isfinite(ir_t1): Ir_stage[s]["t1"].append(ir_t1)
            if np.isfinite(id_t): Id_stage[s]["t"].append(id_t)
            if np.isfinite(id_t1): Id_stage[s]["t1"].append(id_t1)
            if np.isfinite(ig_t): Ig_stage[s]["t"].append(ig_t)
            if np.isfinite(ig_t1): Ig_stage[s]["t1"].append(ig_t1)

print(f"Collected {len(Ir_t)} I_r(t) and {len(Ir_t1)} I_r(t+1) samples.")
print(f"Collected {len(Id_t)} I_d(t) and {len(Id_t1)} I_d(t+1) samples.")
print(f"Collected {len(Ig_t)} I_g(t) and {len(Ig_t1)} I_g(t+1) samples.")

def plot_kde(data, label, color, linestyle="-"):
    kde = gaussian_kde(data)
    xs = np.linspace(*np.percentile(data, [1, 99]), 400)
    plt.plot(xs, kde(xs), color=color, lw=2.2, linestyle=linestyle, label=label)

def plot_density(before, after, xlabel, title, filename):
    before = np.asarray([v for v in before if np.isfinite(v)], dtype=float)
    after  = np.asarray([v for v in after  if np.isfinite(v)], dtype=float)
    if len(before) == 0 or len(after) == 0:
        print(f"Skipping {title} — insufficient data.")
        return

    plt.figure(figsize=(7.5, 5.5))
    plot_kde(before, "t", "blue", "-")
    plot_kde(after, "t+1", "red", "--")
    plt.xlabel(xlabel)
    plt.ylabel("Density")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    if save_fig:
        plt.savefig(os.path.join(base_folder, filename), dpi=150)
        print(f"Saved: {filename}")
    plt.show()

def plot_density_separate(before, after, xlabel, title_prefix, filename_prefix):
    before = np.asarray([v for v in before if np.isfinite(v)], dtype=float)
    after  = np.asarray([v for v in after  if np.isfinite(v)], dtype=float)
    if len(before) == 0 or len(after) == 0:
        print(f"Skipping {title_prefix} — insufficient data.")
        return

    for data, label, color, ls, suffix in [
        (before, "t", "blue", "-", "t"),
        (after, "t+1", "red", "--", "t1")
    ]:
        plt.figure(figsize=(7.5, 5.5))
        plot_kde(data, label, color, ls)
        plt.xlabel(xlabel)
        plt.ylabel("Density")
        plt.title(f"{title_prefix} — {label}")
        plt.tight_layout()
        if save_fig:
            plt.savefig(os.path.join(base_folder, f"{filename_prefix}_{suffix}.png"), dpi=150)
            print(f"Saved: {filename_prefix}_{suffix}.png")
        plt.show()

# ---- Pooled ----
plot_density(Ir_t, Ir_t1, "I_r", r"$I_r$ — R&D Investment (all stages)", "Ir_kde_all.png")
plot_density_separate(Id_t, Id_t1, "I_d", r"$I_d$ — Dirty Investment (all stages)", "Id_kde_all")
plot_density(Ig_t, Ig_t1, "I_g", r"$I_g$ — Green Investment (all stages)", "Ig_kde_all.png")

# ---- Per-tech-stage ----
for s, name in stages.items():
    plot_density(Ir_stage[s]["t"], Ir_stage[s]["t1"], "I_r",
                 rf"$I_r$ — R&D Investment within {name}", f"Ir_kde_stage{s}.png")
    plot_density_separate(Id_stage[s]["t"], Id_stage[s]["t1"], "I_d",
                          rf"$I_d$ — Dirty Investment within {name}", f"Id_kde_stage{s}")
    plot_density(Ig_stage[s]["t"], Ig_stage[s]["t1"], "I_g",
                 rf"$I_g$ — Green Investment within {name}", f"Ig_kde_stage{s}.png")
```
