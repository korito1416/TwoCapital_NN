```python
# Function to read values from a file
def read_values(file_path):
    values = []
    with open(file_path, 'r') as file:
        for line in file:
            values.append(float(line.strip()))
    return values

Name=["Emissions","RD","Id","Ig","TechJumpProb","DmgJumpProb", "T_simulation"]
for txt_name in Name:
    # File paths 5, -1. -3
    # Cap_damage/TwoCapital_NN/output/A_g_prime_1_Delta_0.01_LR_piecewiseconstant_10e-5,10e-5,10e-5,10e-5_128_neurons_32_#HiddenLayer_4_logxi_-3.0_logximax_5.0_num_iterations600000
    file_path1 = FolderName+f'/final_output/pre_tech_pre_damage/log_xi_idx_3/{txt_name}.txt'
    file_path2 = FolderName+f'/final_output/pre_tech_pre_damage/log_xi_idx_2/{txt_name}.txt'
    file_path3 = FolderName+f'/final_output/pre_tech_pre_damage/log_xi_idx_1/{txt_name}.txt'
    file_path4 = FolderName+f'/final_output/pre_tech_pre_damage/log_xi_idx_0/{txt_name}.txt'
  
    # Read values from the files
    values1 = read_values(file_path1)
    values2 = read_values(file_path2)
    values3 = read_values(file_path3)
    values4 = read_values(file_path4)
    values1 = values1[:720]
    values2 = values2[:720]
    values3 = values3[:720]
    values4 = values4[:720]
    # Indices for the values
    indices1 = [x / 12 for x in range(len(values1))]
    indices2 = [x / 12 for x in range(len(values2))]
    indices3 = [x / 12 for x in range(len(values3))]
    indices4 = [x / 12 for x in range(len(values4))]

    # Plotting the values from both files
    plt.figure(figsize=(10, 6))

    plt.plot(indices1, values1,  color='r', label=r'$\xi$ = $\infty$',linewidth=5)
    plt.plot(indices2, values2,  color='g', label=r'$\xi$ = 0.3',linewidth=5)
    plt.plot(indices3, values3,  color='b', label=r'$\xi$ = 0.1',linewidth=5)
    plt.plot(indices4, values4,  color='purple', label=r'$\xi$ = 0.075',linewidth=5)
    
    if txt_name == "RD":
        plt.title(r'R&D investment as % of Output $(I_k/Y)$',fontsize=16)
    elif txt_name =="Id":
        plt.title(r'Dirty investment $(I_d)$',fontsize=16)
    elif txt_name =="Ig":
        plt.title(r'Green investment $(I_g)$',fontsize=16)
    elif txt_name =="TechJumpProb":
        plt.title('Distorted probability of a Technology jump',fontsize=16)
    elif txt_name =="DmgJumpProb":
        plt.title('Distorted probability of a Damage jump',fontsize=16)
    elif txt_name =="TechJumpIntensity":
        plt.title('Distorted Intensity of a Technology jump',fontsize=16)
    elif txt_name =="DmgJumpIntensity":
        plt.title('Distorted Intensity of a Damage jump',fontsize=16)    
        
    else:
        plt.title(f'{txt_name}',fontsize=16)
    plt.xlabel('Year',fontsize=16)
    plt.ylabel('Value',fontsize=16)
    plt.xticks(fontsize=14)  
    plt.yticks(fontsize=14)  
    plt.legend(fontsize=14)  
    plt.show()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def get_log_ticks(data):
    """
    Given an array-like of positive values, compute tick locations as powers of 10.
    Returns:
      ticks: the locations (floats)
      tick_labels: the corresponding labels as strings, e.g., '$10^{-1}$'
    """
    # Ensure data is positive for logarithmic scale
    positive_data = data[data > 0]
    if len(positive_data) == 0:
        return [1], [r'$10^{0}$']
    
    min_exp = int(np.floor(np.log10(positive_data.min())))
    max_exp = int(np.ceil(np.log10(positive_data.max())))
    ticks = [10**i for i in range(min_exp, max_exp + 1)]
    tick_labels = [rf'$10^{{{i}}}$' for i in range(min_exp, max_exp + 1)]
    return ticks, tick_labels

# Define the parent folder and subfolders
foldername = "/project/lhansen/Cap_NN_oldVersion/TwoCapital_NN_with_Scalers/output/OldResults_Delta_0.01_LR_piecewiseconstant_40e-5,40e-5,40e-5,40e-5_128_neurons_32_#HiddenLayer_4_logxi_-3.0_logximax_5.0_num_iterations2000000"
subfolders = [
    'post_tech_post_damage',
    'pre_tech_post_damage',
    'post_tech_pre_damage',
    'pre_tech_pre_damage'
]

# Loop through each subfolder
for sub in subfolders:
    csv_path = os.path.join(foldername, sub, "training_history.csv")
    print(f"Processing {csv_path}...")
    df = pd.read_csv(csv_path)
    
    # Compute test losses
    test_loss_value = df['loss_v'] * df['pv_norm']
    test_loss_dv = df['loss_dv_dY'] * df['marginal_util_consumption_norm']
    
    # Determine dynamic ticks for x (step) and y (loss) for the first plot
    x_ticks, x_tick_labels = get_log_ticks(df['step'])
    y_ticks_value, y_tick_labels_value = get_log_ticks(test_loss_value)
    
    # Plot test loss of value function vs. step
    plt.figure(figsize=(10, 6))
    plt.plot(df['step'], test_loss_value, label=r'$loss\_v \times pv\_norm$', color='blue')
    plt.xlabel('Step')
    plt.ylabel('Test Loss (Value Function)')
    plt.title(f'{sub} - Test Loss of Value Function')
    plt.xscale('log')
    plt.yscale('log')
    plt.xticks(x_ticks, x_tick_labels)
    plt.yticks(y_ticks_value, y_tick_labels_value)
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.savefig(os.path.join(foldername, sub, 'test_loss_value_function.png'))
    plt.show()
    
    # Determine dynamic ticks for x (step) and y (loss) for the second plot
    x_ticks, x_tick_labels = get_log_ticks(df['step'])
    y_ticks_dv, y_tick_labels_dv = get_log_ticks(test_loss_dv)
    
    # Plot test loss of derivative vs. step
    plt.figure(figsize=(10, 6))
    plt.plot(df['step'], test_loss_dv, label=r'$loss\_dv\_dY \times marginal\_util\_consumption\_norm$', color='red')
    plt.xlabel('Step')
    plt.ylabel('Test Loss (Derivative)')
    plt.title(f'{sub} - Test Loss of Derivative')
    plt.xscale('log')
    plt.yscale('log')
    plt.xticks(x_ticks, x_tick_labels)
    plt.yticks(y_ticks_dv, y_tick_labels_dv)
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.savefig(os.path.join(foldername, sub, 'test_loss_derivative.png'))
    plt.show()
```
